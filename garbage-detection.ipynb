```python
import os
import yaml

# --- CẤU HÌNH ---
DATASET_DIR = '/kaggle/input/detection-dataset/detection-dataset' 


CLASS_NAMES = [
    'battery',      # 0
    'biological',   # 1
    'cardboard',    # 2
    'clothes',      # 3
    'glass',        # 4
    'metal',        # 5
    'paper',        # 6
    'plastic',      # 7
    'shoes',        # 8
    'trash'         # 9
]

# Tạo nội dung file yaml
data_yaml = {
    'path': DATASET_DIR,
    'train': 'train/images',
    'val': 'val/images',
    'test': 'test/images',
    'nc': len(CLASS_NAMES),  # Tự động đếm (sẽ bằng 10)
    'names': CLASS_NAMES
}

# Ghi ra file
with open('data.yaml', 'w') as f:
    yaml.dump(data_yaml, f)

print(f"Đã tạo file data.yaml với {len(CLASS_NAMES)} class:")
print(CLASS_NAMES)
```

```python
!wget https://github.com/ultralytics/assets/releases/download/v8.3.0/yolo12n.pt
```

```python
from ultralytics import YOLO

print("Đang tải mô hình YOLO12n ...") 
model = YOLO('yolo12n.pt') 

results = model.train(
    data='data.yaml',      
    imgsz=640,
    epochs=100,
    batch=16,             
    device=0,

    lr0=0.001,              
    patience=30,            
    cos_lr=True,
            

    augment=True,
    hsv_s=0.7,
    hsv_v=0.4,
    
    translate=0.1,          # Dịch chuyển nhẹ
    scale=0.5,              # Zoom: OK

    mosaic=1.0,             
    mixup=0.1,              
    close_mosaic=10,       
    
    workers=4,              
    verbose=True,
    project='garbage_project', 
    name='train_run'
)
```
